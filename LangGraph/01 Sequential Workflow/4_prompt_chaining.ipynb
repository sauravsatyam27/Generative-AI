```python
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
model = ChatGoogleGenerativeAI(
    model = 'gemini-2.5-flash'
)
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    blog: str
```

```python
def create_outline(state: BlogState) -> BlogState:

    # fetch title
    title = state['title']

    # call LLM for generation Outline
    prompt = f'Generate an Outline for the topic {title}'

    answer = model.invoke(prompt).content

    # Update State
    state['outline'] = answer

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:

    # fetch title
    title = state['title']
    outline = state['outline']



    # call LLM for generation Outline
    prompt = f'Generate an detiled blog for the topic - {title} using the following outline \n {outline}'

    result = model.invoke(prompt).content

    # Update State
    state['blog'] = result

    return state    
```

```python
graph = StateGraph(BlogState)

#nodes

graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)

## Add the edges

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', END)


## Workflow

work_flow = graph.compile()
```

```python
initial_state = {'title': 'Rise of AI in India'}

final_state = work_flow.invoke(initial_state)

print(final_state['blog'])
```
