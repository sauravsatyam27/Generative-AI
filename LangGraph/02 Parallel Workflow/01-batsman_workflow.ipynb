```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):

    runs:  int
    balls: int
    fours: int
    sixes: int

    sr: float
    bpb: float
    boundary_percentage: float

    summary : str
```

```python
def calculate_sr(state:BatsmanState) -> BatsmanState:
    runs = state['runs']
    balls = state['balls']

    strikeRate = (runs / balls) * 100

   

    return {
        'sr': strikeRate
    }
```

```python
def calculate_bpb(state:BatsmanState) -> BatsmanState:
    balls = state['balls']
    fours = state['fours']   
    sixes = state['sixes'] 
     

    ballBoundary = balls / (fours + sixes)

    return {
        'bpb' : ballBoundary
    }

    return state 
```

```python
def calculate_boundary_percentage(state:BatsmanState) -> BatsmanState:

    balls = state['balls']
    fours = state['fours']   
    sixes = state['sixes'] 
    runs = state['runs']

    fourRuns = fours * 4
    sixRuns = sixes * 6

    boundaryRuns = fourRuns + sixRuns

    bdryPercent = (boundaryRuns / runs) * 100


    return {
        'boundary_percentage' :bdryPercent
    }

    
```

```python
def summary(state:BatsmanState) -> BatsmanState:

    balls = state['balls']
    fours = state['fours']   
    sixes = state['sixes'] 
    runs = state['runs']
    sr = state['sr']
    bpb = state['bpb']
    boundary_percentage = state['boundary_percentage']

    summary = (
        f"The batsman scored {runs} runs in {balls} balls. "
        f"He hit {fours} fours and {sixes} sixes. "
        f"His strike rate was {sr:.2f}. "
        f"He scored {bpb:.2f} balls per boundary. "
        f"{boundary_percentage:.2f}% of his runs came from boundaries."
    )


    return {
        'summary': summary
    }
```

```python
graph = StateGraph(BatsmanState)

graph.add_node('calculate_sr', calculate_sr)
graph.add_node('calculate_bpb', calculate_bpb)
graph.add_node('calculate_boundary_percentage', calculate_boundary_percentage)
graph.add_node('summary', summary)


## Add the edges

graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'calculate_bpb')
graph.add_edge(START, 'calculate_boundary_percentage')

graph.add_edge('calculate_sr', 'summary')
graph.add_edge('calculate_bpb', 'summary')
graph.add_edge('calculate_boundary_percentage', 'summary')

graph.add_edge('summary', END)


work_flow = graph.compile()
```

```python
inital_state = {
    'runs': 100,
    'balls' : 30,
    'fours' : 10,
    'sixes' : 6
}

work_flow.invoke(inital_state)
```
