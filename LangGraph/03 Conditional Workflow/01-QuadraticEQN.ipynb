```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict, Literal
```

```python
class QuadState(TypedDict):

    a : int
    b : int
    c : int

    equation : str
    discriminant : float
    result : str
```

```python
def show_equation(state : QuadState):

    equation = f'{state["a"]}x2 {state["b"]}x {state["c"]}'

    return{
        "equation": equation
    }


def calculate_discriminant(state : QuadState):

    discriminant = state["b"]**2 - (4* state["a"] * state["c"])

    return {
        "discriminant" : discriminant
    }
```

```python
def real_root(state: QuadState):

    root1 = (-state["b"] + state["discriminant"] ** 0.5) / (2 * state["a"])
    root2 = (-state["b"] - state["discriminant"] ** 0.5) / (2 * state["a"])

    result = f'The Two root are : {root1} and {root2}'

    return {
        "result" : result
    }


def repetead_root(state: QuadState):

    root = (-state["b"]) / (2 * state["a"])

    result = f'The Two root are : {root} and {root}'

    return {
        "result" : result
    }


def no_real_root(state: QuadState):

   # root = (-state["b"]) / (2 * state["a"])

    result = f'No real Roots'

    return {
        "result" : result
    }


def check_condition(state : QuadState) -> Literal ["real_root","repetead_root","no_real_root"]:
    if state["discriminant"] > 0:
        return "real_root"

    elif state["discriminant"] == 0:
        return "repetead_root"

    else :
        return "no_real_root"
```

```python
graph = StateGraph(QuadState)

graph.add_node('show_equation', show_equation)
graph.add_node('calculate_discriminant',calculate_discriminant)
graph.add_node('real_root',real_root)
graph.add_node('repetead_root',repetead_root)
graph.add_node('no_real_root',no_real_root)

graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation', 'calculate_discriminant')


graph.add_conditional_edges('calculate_discriminant',check_condition)

graph.add_edge('real_root',END)
graph.add_edge('repetead_root',END)
graph.add_edge('no_real_root',END)


graph.add_edge('calculate_discriminant', END)

workflow = graph.compile()
```

```python
initial_state = {
    'a' : -4,
    'b' : 5,
    'c' : 2
}

workflow.invoke(initial_state)
```
